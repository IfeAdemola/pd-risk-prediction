```python
import pandas as pd

pd_cases = pd.read_csv(
    "../data/exploratory/outcomes/ppmi_pd_conversion_candidates.csv"
)

pd_cases["pd_case_type"].value_counts()
```

```python
def show_pd_examples(
    df,
    category,
    n=10,
):
    print("\n")
    print("=" * 80)
    print(category)
    print("=" * 80)

    print(
        df[
            df["pd_case_type"] == category
        ][
            [
                "PATNO",
                "sequence",
                "n_visits",
                "first_pd_visit",
                "first_pd_date",
                "final_diagnosis",
                "ever_msa",
                "ever_dlb",
                "death",
            ]
        ]
        .head(n)
        .to_string(index=False)
    )

```

```python

show_pd_examples(
    pd_cases,
    "persistent_pd",
)

show_pd_examples(
    pd_cases,
    "pd_reversal",
)

show_pd_examples(
    pd_cases,
    "pd_competing_diagnosis",
)

show_pd_examples(
    pd_cases,
    "single_pd_visit",
)

show_pd_examples(
    pd_cases,
    "pd_late_conversion",
)
```
