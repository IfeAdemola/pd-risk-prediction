#### DAT SANITY CHECK

**Cell 1 — Load the curated longitudinal DAT cohort**

```python
# ============================================================
# DAT INVESTIGATION — CURATED VS RAW LONGITUDINAL DATA
# ============================================================

curated_dat = pd.read_csv(
    PROJECT_ROOT / "data/processed/cohort/ppmi_risk_cohort_longitudinal.csv"
)

dat = pd.read_csv(PROJECT_ROOT / "data/raw/ppmi/Xing_Core_Lab_-_Quant_SBR_28Jul2026.csv"
)

raw_dat_cohort = pd.read_csv(PROJECT_ROOT / "output/manifest/ppmi_risk_cohort_visit_manifest.csv")

print(f"Curated DAT rows:          {len(curated_dat):,}")
print(f"Curated participants:      {curated_dat['PATNO'].nunique():,}")
print(f"Curated visits:            {curated_dat['EVENT_ID'].nunique():,}")
```

**Cell 2 — Define the DAT feature set**

```python
# ============================================================
# DAT FEATURE SET
# ============================================================

DAT = [
    "MIA_CAUDATE_L",
    "MIA_CAUDATE_R",
    "MIA_PUTAMEN_L",
    "MIA_PUTAMEN_R",
    "MIA_STRIATUM_L",
    "MIA_STRIATUM_R",
]

missing = [col for col in DAT if col not in curated_dat.columns]

if missing:
    print("Missing DAT features:", missing)
else:
    print("✓ All DAT features are present")
```

**Cell 4 — Compare visit distributions**

```python
# ============================================================
# VISIT-LEVEL COMPARISON
# ============================================================

curated_visits = (
    curated_dat
    .groupby("EVENT_ID")["PATNO"]
    .nunique()
    .rename("curated_n")
)

raw_visits = (
    raw_dat_cohort
    .groupby("EVENT_ID")["PATNO"]
    .nunique()
    .rename("raw_n")
)

visit_comparison = (
    pd.concat([curated_visits, raw_visits], axis=1)
    .fillna(0)
    .astype(int)
)

visit_comparison["difference"] = (
    visit_comparison["curated_n"]
    - visit_comparison["raw_n"]
)

visit_comparison
```

**Cell 5 — Compare percentages**

```python
# ============================================================
# VISIT-LEVEL COVERAGE
# ============================================================

visit_comparison_pct = visit_comparison.copy()

visit_comparison_pct["curated_pct"] = (
    100 * visit_comparison_pct["curated_n"]
    / len(cohort_patients)
)

visit_comparison_pct["raw_pct"] = (
    100 * visit_comparison_pct["raw_n"]
    / len(cohort_patients)
)

visit_comparison_pct[
    ["curated_n", "curated_pct", "raw_n", "raw_pct", "difference"]
]
```

```python
RAW_DAT = [
    "CAUDATE_L_REF_CWM",
    "CAUDATE_R_REF_CWM",
    "PUTAMEN_L_REF_CWM",
    "PUTAMEN_R_REF_CWM",
    "STRIATUM_L_REF_CWM",
    "STRIATUM_R_REF_CWM",
]
```

```python
# ============================================================
# INDIVIDUAL-LEVEL SC → BL CHECK
# ============================================================

example_patients = sorted(
    set(
        raw_dat_cohort.loc[
            raw_dat_cohort["EVENT_ID"].eq("SC"),
            "PATNO"
        ]
    )
)[:10]

for patno in example_patients:
    print(f"\nPATNO {patno}")

    print("Raw:")
    print(
        raw_dat_cohort.loc[
            raw_dat_cohort["PATNO"].eq(patno),
            ["PATNO", "EVENT_ID"] + RAW_DAT
        ]
    )

    print("Curated:")
    print(
        curated_dat.loc[
            curated_dat["PATNO"].eq(patno),
            ["PATNO", "EVENT_ID"] + DAT
        ]
    )
```
